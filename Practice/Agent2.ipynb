```python
from typing import TypedDict,Dict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
 name: str
 value: list[int] 
 result:str
```

```python
def proccess_value(state:AgentState)->AgentState:
    state['result']=f'Hello {state["name"]},Your scores is{sum(state['value'])}'
    return state
```

```python
graph= StateGraph(AgentState)
graph.add_node("proccess_value",proccess_value)
graph.set_entry_point('proccess_value')
graph.set_finish_point('proccess_value')

app =graph.compile()
```

```python
from IPython.display import Image,display
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
result=app.invoke({'name':'Deep','value':[10,20,30]})
```
